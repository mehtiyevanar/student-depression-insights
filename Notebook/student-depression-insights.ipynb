Source et contexte des données
</p>

Le jeu de données utilisé est le **Student Depression Dataset**, disponible sur la plateforme Kaggle à l’adresse suivante : [Student Depression Dataset – Kaggle](https://www.kaggle.com/datasets/hopesb/student-depression-dataset).

Il s’agit d’un jeu de données tabulaire au format CSV portant sur la santé mentale des étudiants, avec un intérêt particulier pour l’étude de la dépression et des facteurs pouvant y être associés.

Chaque ligne du jeu de données correspond à un étudiant, tandis que chaque colonne représente une caractéristique individuelle, académique, comportementale ou psychologique. Le jeu de données regroupe notamment des informations relatives au profil sociodémographique des étudiants, à leur environnement académique, à leurs habitudes de vie, ainsi qu’à certains indicateurs liés au bien-être mental.

Ce jeu de données permet d’étudier les relations possibles entre la dépression étudiante et plusieurs facteurs explicatifs tels que la pression académique, la satisfaction dans les études, la durée du sommeil, les habitudes alimentaires, le stress financier ou encore les antécédents familiaux liés à la santé mentale. Il offre ainsi une base pertinente pour mener une analyse exploratoire des données, produire des visualisations interprétables et construire éventuellement des analyses statistiques ou économétriques.

Dans le cadre de cette analyse, la variable liée à la dépression constitue la variable d’intérêt principale. Les autres variables seront mobilisées comme variables explicatives ou variables de segmentation afin d’identifier les profils d’étudiants les plus exposés au risque de dépression.

Il convient toutefois de souligner que ce jeu de données porte sur un sujet sensible : la santé mentale. Les résultats obtenus devront donc être interprétés avec prudence. L’objectif de l’analyse n’est pas d’établir un diagnostic médical, mais d’identifier des tendances statistiques et des associations entre variables. Toute conclusion devra rester limitée au périmètre du jeu données et être formulée dans une perspective analytique, préventive et non clinique.

```python
# Suppression des observations dont l'effectif est inférieur ou égal à 10
df = df[df.groupby("Âge")["Âge"].transform("size") > 10]

# Vérification des observations par "Âge" après suppression
df["Âge"].value_counts()
```

```python
# Répartition des observations par "Ville"
df["Ville"].value_counts()
```

```python
# Suppression des observations comptant moins de 400 effectifs
effectif_ville = df["Ville"].value_counts()
villes_a_supprimer = effectif_ville[effectif_ville < 400].index

df = df[~df["Ville"].isin(villes_a_supprimer)]

# Vérification des observations par "Ville" après suppression
df["Ville"].value_counts()
```
